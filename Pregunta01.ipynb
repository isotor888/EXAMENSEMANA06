```python
# Programa un menú con dos opciones
# Apellidos y Nombres: Soto Rodriguez Ismael Gilver
# fecha: 27/04/2025
while True:
    print("MENU PRINCIPAL")
    print("a. OPCION CONTAR CONSONANTES SIN VOCALES")
    print("b. SALIR")
    OPCIONES = input("ELIGE UNA OPCION (a/b): ").lower()
    if OPCIONES == "a":
        texto = input("INGRESA UNA CADENA DE TEXTO: ")
        def CONTARCONSONANTES(texto):
            VOCALES = "AEIOUaeiou"
            consonantes = 0
            for caracter in texto:
                if caracter.isalpha() and caracter not in VOCALES:
                    consonantes += 1
            return consonantes
        resultado = CONTARCONSONANTES(texto)
        print(f"LA CANTIDAD DE CONSONANTES (SIN CONTAR VOCALES) ES: {resultado}")

    elif OPCIONES == "b":
            print("USTED SALIO DEL PROGRAMA")
            break
    else:
            print("OPCION NO VALIDA")
```
