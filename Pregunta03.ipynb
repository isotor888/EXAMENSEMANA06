```python
# función recursiva para calcular la división de dos números
# Apellidos y Nombres: Soto Rodriguez Ismael Gilver
# fecha: 27/047/2025

def fundivision(w, z):      # 1 (10,2)        ## (5,2)        ### (3,2)       ####(1,2)
    if (w == 0 or z == 0):               # 10  y 2 son diferente de 0     ## 5 y 2 son diferente de 0     ### 3 y 2 son diferente de 0      #### 1 y 2 son diferentes de 0
        return 0
    else:
        return 1 + fundivision(w-z,z)     # 1 + fundivision(5,2)         ## 1 + fundivision(3,2)        ### 1 + fundivision (1,2)         #### 1 + fundiv(-1,2)
                                                              # +3 +2 +1 -1 -------> 5
a = 10
b = 2
resultado = fundivision(a, b)
print(f"{a} dividido entre {b} es: {resultado}")
```
